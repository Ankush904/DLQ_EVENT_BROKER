```python
from datetime import date
from pathlib import Path
from rich import print
import json
```

```python


PROJECT_DIR = Path.cwd()
OUTPUT_FOLDER_NAME = "messages"


def ordinal(day: int) -> str:
    if 11 <= day % 100 <= 13:
        suffix = "th"
    else:
        suffix = {1: "st", 2: "nd", 3: "rd"}.get(day % 10, "th")
    return f"{day}{suffix}"


def get_today_folder() -> Path:
    today = date.today()
    folder_name = f"{ordinal(today.day)} {today.strftime('%B %Y')}"
    return PROJECT_DIR / folder_name


today_folder = get_today_folder()
messages_folder = today_folder / OUTPUT_FOLDER_NAME

file_names = sorted(path.name for path in messages_folder.iterdir() if path.is_file())
```

```python
print(file_names)
```

```python
ml_inference_results_list = []
process_callback_list = []
recorded_conversations_list = []

for file_name in file_names:
    if file_name.startswith("ml_inference_results"):
        file_path = messages_folder / file_name
        with open(file_path, "r") as file:
            ml_inference_results_list.extend(json.load(file))
    elif file_name.startswith("process_callback"):
        file_path = messages_folder / file_name
        with open(file_path, "r") as file:
            process_callback_list.extend(json.load(file))
    elif file_name.startswith("recorded_conversations"):
        file_path = messages_folder / file_name
        with open(file_path, "r") as file:
            recorded_conversations_list.extend(json.load(file))
```
